```python
df= pd.read_csv("../mizban_orders.csv")
```

```python
X= df.drop(columns=["orders","date"])
y= df[["orders"]]
order_count=df["orders"].value_counts()
mask= y.isin(order_count[order_count > 2].index)

print(y[["orders"]])
print(y["orders"])

df_filtered = df[mask.values] 

X_filtered = df_filtered.drop(columns=["orders", "date"])
y_filtered = df_filtered["orders"]

print(f"number of data befor filter: {len(df)}")
print(f"number of data after filter: {len(df_filtered)}")
```

```python
def split_dataset(X,y, test_size=0.2, random_state=42):
    random_indeces= np.random.permutation(len(X))
    
    test_set_size= int(len(random_indeces) * test_size)
    
    train_size_indeces= random_indeces[test_set_size:]
    test_size_indeces= random_indeces[:test_set_size]
    
    X_train, X_test= X.iloc[train_size_indeces], X.iloc[test_size_indeces]
    y_train, y_test= y.iloc[train_size_indeces], y.iloc[test_size_indeces]
    
    return X_train, X_test, y_train, y_test
```
